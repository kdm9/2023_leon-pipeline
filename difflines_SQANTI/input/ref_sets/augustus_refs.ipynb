```python
accessions= ['at6137','at6923','at6929','at7143','at8285','at9104','at9336','at9503','at9578','at9744','at9762','at9806','at9830','at9847','at9852','at9879','at9883','at9900','Araport']
paths=["busco_Araport11_20221221.txt",
       "arat_housekeeping.txt"]
ref=["busco","housekeeping"]

for path, ref in zip(paths, ref):
    print(ref)
    all_orthogroups = open("orthogroups.txt", 'r')
    Orthogroups = all_orthogroups.readlines()

    ref_set=open(path, 'r')
    Refgenes=ref_set.readlines()
    refgenes = []
    for gene in Refgenes:
        refgenes.append(gene.strip().split(".")[0])
    
    mappings = {}
    for acc in accessions:
        mappings[acc]=[]
            
    # Strips the newline character
    for group in Orthogroups:
        a2g = group.strip().split(": ")[1].split(" ")
        OGs.append(group.strip().split(": ")[0])
        #print(a2g)
        included=0
        for c in a2g:
            if c.split("_")[1].split(".")[0] in refgenes:
                included=1
            if c.split("_")[0]=="Araport":
                all_Araport.append(c.split("_")[1].split(".")[0])                

        if included == 1:
            for c in a2g:
                #print(c)
                #print(c.split("_")[1])
                acc = c.split("_")[0]
                gene = "_".join(c.split("_")[:2])
                try:
                    mappings[acc].append(gene)
                except:
                    continue
                
    for acc in mappings.keys():
        if acc != "Araport":
            with open(f'new_ids/{ref}_{acc}.txt', 'a') as out:
                for gene in mappings[acc]:
                    out.write(f'{gene}\n')
        else:
            with open(f'new_ids/{ref}_col0.txt', 'a') as out:
                for gene in mappings[acc]:
                    out.write(f'{gene}\n')
    
```
