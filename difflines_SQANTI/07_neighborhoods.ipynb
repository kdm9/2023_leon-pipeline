```python
import pandas as pd
```

```python
all_nei=pd.DataFrame()
for i in range(1,6):
    nei=pd.read_csv(f"input/neighborhoods/final_all_neighborhoods.chr{i}.bed", sep="\t", header=None)
    all_nei=pd.concat([all_nei,nei])
all_nei.columns=["chrom","genomic_start_coord","genomic_end_coord","neighborhood_id"]
```

```python
all_nei["accession"]=all_nei["chrom"].str.split("_").str[0]
all_nei=all_nei.loc[all_nei["accession"]!="at6137"]
```

```python
genes_in_neighborhoods=pd.DataFrame()
for acc in all_nei["accession"].unique().tolist():
    genes=pd.read_csv(f"output/06/{acc}/{acc}-hodgepodgemerged_cat+subcat_iso-rs.bed", sep="\t")
    genes=genes[['chrom', 'genomic_start_coord', 'genomic_end_coord', 'strand','hodgepodge_gene', 'type','read_support', 'isoforms']]
    genes["read_support"]=genes.groupby("hodgepodge_gene")["read_support"].transform(sum)
    genes["isoforms"]=genes.groupby("hodgepodge_gene")["isoforms"].transform(sum)
    genes.drop_duplicates(inplace=True)
    
    acc_df=pd.DataFrame()
    for chrom in ["chr1","chr2","chr3","chr4","chr5"]:
        chrom_nei=all_nei.loc[all_nei["chrom"]==f"{acc}_1_{chrom}"]
        genes_chrom=genes.loc[genes["chrom"]==f"{acc}_1_{chrom}"]
        #genes["neighborhood"]="empty"

        for start, end, nei in zip(chrom_nei["genomic_start_coord"].tolist(), chrom_nei["genomic_end_coord"].tolist(), chrom_nei["neighborhood_id"].tolist()):
            x=genes_chrom.loc[(genes_chrom["genomic_start_coord"]>=start)&(genes_chrom["genomic_end_coord"]<=end)].copy()
            x["neighborhood"]=nei
            acc_df=pd.concat([acc_df,x])
    genes_in_neighborhoods=pd.concat([genes_in_neighborhoods,acc_df])

        #print(genes)
```

```python
genes_in_neighborhoods.to_csv("output/07/genes_in_neighborhoods.bed", sep="\t",index=False)
```
