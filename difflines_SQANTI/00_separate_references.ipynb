### Housekeeping

```python
housekeeping_list=[]

file = open("input/ref_sets/arat_housekeeping.txt", 'r')
Lines = file.readlines()

for line in Lines:
    gene=line.strip().split(".")[0]
    housekeeping_list.append(gene)
    
housekeeping_list=list(set(housekeeping_list))
```

```python
accessions= ['at6923','at6929','at7143','at8285','at9104','at9336','at9503','at9578','at9744','at9762','at9806','at9830','at9847','at9852','at9879','at9883','at9900','Araport']
paths=["/ebio/abt6_projects7/dliso/dlis/leon_pipeline/difflines_SQANTI/input/ref_sets/Orthogroups_26-09-23.txt"]

ref=["housekeeping"]

for path, ref in zip(paths, ref):
    file = open(path, 'r')
    Lines = file.readlines()

    mappings = {}
    for acc in accessions:
        mappings[acc]=[]

    # Strips the newline character
    for line in Lines:
        house_og=0
        a2g = line.strip().split(": ")[1].split(" ")
        for c in a2g:
            # print(c.split("_")[1])
            acc = c.split("_")[0]
            if c.split("_")[0] =="Araport":
                gene=c.split("_")[1].split(".")[0]
            else:
                gene=c.rsplit("_",1)[0]
            if gene in housekeeping_list:
                house_og+=1
                break
        
        if house_og==1:
            for c in a2g:
                acc = c.split("_")[0]
                if c.split("_")[0] =="Araport":
                    gene=c.split("_")[1].split(".")[0]
                else:
                    gene=c.rsplit("_",1)[0]
                    
                try:
                    mappings[acc].append(gene)
                except:
                    continue

    for acc in mappings.keys():
        if acc != "Araport":
            with open(f'input/ref_sets/new_ids/{ref}_{acc}.txt', 'a') as out:
                for gene in mappings[acc]:
                    out.write(f'{gene}\n')
        else:
            with open(f'input/ref_sets/new_ids/{ref}_col0.txt', 'a') as out:
                for gene in mappings[acc]:
                    out.write(f'{gene}\n')
```
